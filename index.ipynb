# Introduction to Python Workshop
## 12 Self-Contained Sessions with Practical Projects

Welcome to the Introduction to Python workshop! This course teaches Python fundamentals through hands-on, practical projects. Each session is designed to be completed in 30 minutes and results in a working tool you can use and adapt.

---

## 🎯 Course Goals

By the end of this workshop, you will be able to:
- Write Python programs using core language features
- Work with different data types and structures
- Read and write files for data persistence
- Analyze data with pandas
- Create visualizations with matplotlib
- Build complete end-to-end data analysis projects

---

## 📚 Course Structure

### Part 1: Python Fundamentals (Sessions 1-4)
Learn the building blocks of Python programming.

#### [Session 1: Calculator & Variables](session_01_calculator_and_variables.ipynb)
**What you'll learn:** Basic arithmetic, variables, and assignment  
**Practical project:** Budget calculator  
**Time:** 30 minutes  
**Prerequisites:** None

#### [Session 2: Strings and Text Manipulation](session_02_strings_and_text.ipynb)
**What you'll learn:** Working with text, string methods, f-strings  
**Practical project:** Email template generator  
**Time:** 30 minutes  
**Prerequisites:** Session 1

#### [Session 3: Lists and Basic Data Collections](session_03_lists_and_collections.ipynb)
**What you'll learn:** Lists, indexing, slicing, list methods  
**Practical project:** Reading list manager  
**Time:** 30 minutes  
**Prerequisites:** Sessions 1-2

#### [Session 4: Loops and Iteration](session_04_loops_and_iteration.ipynb)
**What you'll learn:** For loops, while loops, iterating through collections  
**Practical project:** Grade analyzer  
**Time:** 30 minutes  
**Prerequisites:** Sessions 1-3

### Part 2: Intermediate Concepts (Sessions 5-8)
Build more complex programs with conditionals, dictionaries, and functions.

#### [Session 5: Conditionals and Decision Making](session_05_conditionals_and_decisions.ipynb)
**What you'll learn:** If/elif/else statements, comparison operators, logical operators  
**Practical project:** Decision-making tool (bike or drive based on conditions)  
**Time:** 30 minutes  
**Prerequisites:** Sessions 1-4

#### [Session 6: Dictionaries and Structured Data](session_06_dictionaries.ipynb)
**What you'll learn:** Dictionaries, key-value pairs, nested structures  
**Practical project:** Searchable contact database  
**Time:** 30 minutes  
**Prerequisites:** Sessions 1-5

#### [Session 7: Functions and Code Reusability](session_07_functions.ipynb)
**What you'll learn:** Defining functions, parameters, return values, documentation  
**Practical project:** Unit converter library (temperature, distance, weight)  
**Time:** 30 minutes  
**Prerequisites:** Sessions 1-6

#### [Session 8: Working with Files](session_08_files.ipynb)
**What you'll learn:** Reading/writing text files, CSV files, file handling best practices  
**Practical project:** Journaling system with search functionality  
**Time:** 30 minutes  
**Prerequisites:** Sessions 1-7

---

### Part 3: Data Analysis (Sessions 9-12)
Learn to analyze and visualize data using pandas and matplotlib.

#### [Session 9: Introduction to Pandas](session_09_pandas_intro.ipynb)
**What you'll learn:** DataFrames, reading CSV files, filtering, basic statistics  
**Practical project:** Book ratings dataset analysis  
**Time:** 30 minutes  
**Prerequisites:** Sessions 1-8  
**New packages:** pandas, numpy

#### [Session 10: Data Cleaning and Transformation](session_10_data_cleaning.ipynb)
**What you'll learn:** Handling missing values, sorting, grouping, creating new columns  
**Practical project:** Clean and prepare messy sales data  
**Time:** 30 minutes  
**Prerequisites:** Sessions 1-9

#### [Session 11: Data Visualization with Matplotlib](session_11_visualization.ipynb)
**What you'll learn:** Line plots, bar charts, scatter plots, customization  
**Practical project:** Personal finance dashboard  
**Time:** 30 minutes  
**Prerequisites:** Sessions 1-10  
**New packages:** matplotlib

#### [Session 12: Putting It All Together](session_12_capstone.ipynb)
**What you'll learn:** Combining all skills in a complete project  
**Practical project:** Fitness tracker analysis with full dashboard  
**Time:** 30 minutes  
**Prerequisites:** Sessions 1-11

## 🚀 Getting Started

### Quick Start Guide

1. **Install Python** (if you haven't already)
   - Download from [python.org](https://www.python.org/downloads/)
   - Version 3.8 or higher recommended

2. **Install required packages**
   ```bash
   pip install -r requirements.txt
   ```

3. **Start with Session 1**
   - Click the link above or open `session_01_calculator_and_variables.ipynb`
   - Sessions 1-8 only need Jupyter (no other packages)
   - Install pandas/matplotlib before Session 9

### Navigation Tips

- **From any session**: Use your file browser to return to this index
- **Self-contained**: Each session can be completed independently
- **Practice cells**: Look for `# Your code here` comments
- **Bonus challenges**: Extra exercises for faster learners

## 💻 Using These Notebooks

### Running Code Cells
- **Execute a cell**: `Shift + Enter` or click the "Run" button
- **Edit a cell**: Click inside it to edit
- **Add new cell**: Click the `+` button in the toolbar

### Keyboard Shortcuts
- `Shift + Enter`: Run cell and move to next
- `Ctrl + Enter`: Run cell and stay
- `A`: Add cell above (in command mode)
- `B`: Add cell below (in command mode)
- `DD`: Delete cell (in command mode)
- `M`: Change to markdown cell
- `Y`: Change to code cell

### Getting Help
- Use `help(function_name)` in any code cell
- Add `?` after a function: `print?`
- Check the official documentation: [docs.python.org](https://docs.python.org/)

## 🔧 Troubleshooting

### Common Issues

**Jupyter won't start**
```bash
jupyter notebook --port 8888
```

**Package not found**
```bash
pip install --upgrade pip
pip install package_name
```

**Kernel issues**
- Try: Kernel → Restart
- Or: Kernel → Restart & Clear Output

**Code cell won't run**
- Check for syntax errors
- Ensure you ran previous cells in order
- Try restarting the kernel

## 👥 About This Course

This workshop was designed for librarians and researchers who want to learn Python through practical, immediately useful projects. Each session builds a real tool that participants can adapt to their own research and workflow needs.

### Design Philosophy
- **Practical over theoretical**: Learn by building useful tools
- **Self-contained sessions**: Join at any point
- **Progressive complexity**: Build skills gradually
- **Real-world applications**: Connect to actual research tasks
- **Immediate value**: Take home working code each session

### Workshop Format
Each 30-minute session follows this structure:
- **5 minutes**: Concept introduction with live examples
- **10 minutes**: Guided practice exercises
- **15 minutes**: Build practical project together
- **Bonus**: Challenge exercises for faster learners
