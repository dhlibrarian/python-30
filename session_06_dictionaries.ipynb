## Part 1: Introduction to Dictionaries (5 minutes)

Dictionaries store data as key-value pairs, like a real dictionary with words and definitions.

```python
# Creating a dictionary
person = {
    "name": "Alice",
    "age": 30,
    "city": "New York"
}

print(person)
```

```python
# Accessing values
print(person["name"])
print(person["age"])

# Safer way with .get()
print(person.get("city"))
print(person.get("email", "Not provided"))  # Default value if key doesn't exist
```

```python
# Empty dictionary
empty_dict = {}
print(empty_dict)
```

## Part 2: Modifying Dictionaries (5 minutes)

```python
# Adding or updating entries
person = {"name": "Alice", "age": 30}
print("Original:", person)

person["email"] = "alice@example.com"  # Add new key
print("After adding email:", person)

person["age"] = 31  # Update existing key
print("After updating age:", person)
```

```python
# Removing entries
person = {"name": "Alice", "age": 30, "city": "NYC"}

del person["city"]  # Remove specific key
print(person)

removed_value = person.pop("age")  # Remove and return value
print(f"Removed: {removed_value}")
print(person)
```

```python
# Checking if key exists
person = {"name": "Alice", "age": 30}

if "name" in person:
    print(f"Name is: {person['name']}")

if "email" not in person:
    print("No email on file")
```

```python
# Dictionary methods
person = {"name": "Alice", "age": 30, "city": "NYC"}

print("Keys:", list(person.keys()))
print("Values:", list(person.values()))
print("Items:", list(person.items()))
```

## Part 3: Looping Through Dictionaries (5 minutes)

```python
# Loop through keys
person = {"name": "Alice", "age": 30, "city": "NYC"}

for key in person:
    print(key)
```

```python
# Loop through values
for value in person.values():
    print(value)
```

```python
# Loop through key-value pairs
for key, value in person.items():
    print(f"{key}: {value}")
```

```python
# Practical example: counting items
fruits = ["apple", "banana", "apple", "cherry", "banana", "apple"]
fruit_count = {}

for fruit in fruits:
    if fruit in fruit_count:
        fruit_count[fruit] += 1
    else:
        fruit_count[fruit] = 1

print("Fruit counts:")
for fruit, count in fruit_count.items():
    print(f"  {fruit}: {count}")
```

## Part 4: Nested Dictionaries and Lists (5 minutes)

```python
# Dictionary with list values
person = {
    "name": "Alice",
    "hobbies": ["reading", "hiking", "coding"],
    "scores": [85, 90, 88]
}

print(f"{person['name']}'s hobbies: {person['hobbies']}")
print(f"First hobby: {person['hobbies'][0]}")
```

```python
# Nested dictionaries
address = {
    "street": "123 Main St",
    "city": "New York",
    "zip": "10001"
}

person = {
    "name": "Alice",
    "address": address
}

print(f"City: {person['address']['city']}")
```

```python
# List of dictionaries
contacts = [
    {"name": "Alice", "phone": "555-0001"},
    {"name": "Bob", "phone": "555-0002"},
    {"name": "Charlie", "phone": "555-0003"}
]

for contact in contacts:
    print(f"{contact['name']}: {contact['phone']}")
```

## Part 5: Contact Database (10 minutes)

Let's build a practical contact management system with search functionality.

```python
# Contact Database - Complete Version

contacts = []

def add_contact(name, phone, email, category="personal"):
    """Add a new contact to the database."""
    contact = {
        "name": name,
        "phone": phone,
        "email": email,
        "category": category
    }
    contacts.append(contact)
    print(f"✓ Added {name} to contacts")

def display_all():
    """Display all contacts."""
    if not contacts:
        print("No contacts in database")
        return
    
    print("\n===== ALL CONTACTS =====")
    for i, contact in enumerate(contacts, 1):
        print(f"\n{i}. {contact['name']} [{contact['category']}]")
        print(f"   Phone: {contact['phone']}")
        print(f"   Email: {contact['email']}")
    print("=" * 25)

def search_contact(search_term):
    """Search for contacts by name."""
    results = []
    search_term = search_term.lower()
    
    for contact in contacts:
        if search_term in contact['name'].lower():
            results.append(contact)
    
    if results:
        print(f"\nFound {len(results)} result(s):")
        for contact in results:
            print(f"\n{contact['name']}")
            print(f"  Phone: {contact['phone']}")
            print(f"  Email: {contact['email']}")
            print(f"  Category: {contact['category']}")
    else:
        print(f"No contacts found matching '{search_term}'")

def filter_by_category(category):
    """Display contacts in a specific category."""
    results = [c for c in contacts if c['category'] == category]
    
    if results:
        print(f"\n===== {category.upper()} CONTACTS =====")
        for contact in results:
            print(f"{contact['name']}: {contact['phone']}")
    else:
        print(f"No {category} contacts found")

def update_contact(name, **kwargs):
    """Update contact information."""
    for contact in contacts:
        if contact['name'].lower() == name.lower():
            for key, value in kwargs.items():
                if key in contact:
                    contact[key] = value
            print(f"✓ Updated {name}")
            return
    print(f"Contact {name} not found")

# Test the database
add_contact("Alice Johnson", "555-0001", "alice@example.com", "work")
add_contact("Bob Smith", "555-0002", "bob@example.com", "personal")
add_contact("Charlie Brown", "555-0003", "charlie@example.com", "work")
add_contact("Diana Prince", "555-0004", "diana@example.com", "personal")

display_all()
```

```python
# Search for contacts
search_contact("alice")
```

```python
# Filter by category
filter_by_category("work")
```

```python
# Update a contact
update_contact("Bob Smith", phone="555-9999", email="newbob@example.com")
search_contact("bob")
```

### Enhanced Version: Contact Statistics

```python
def get_statistics():
    """Display database statistics."""
    if not contacts:
        print("No contacts in database")
        return
    
    # Count by category
    category_counts = {}
    for contact in contacts:
        category = contact['category']
        category_counts[category] = category_counts.get(category, 0) + 1
    
    print("\n===== DATABASE STATISTICS =====")
    print(f"Total contacts: {len(contacts)}")
    print("\nBy category:")
    for category, count in category_counts.items():
        print(f"  {category}: {count}")
    print("=" * 31)

get_statistics()
```
