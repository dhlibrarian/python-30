## Part 2: Variables (10 minutes)

Variables let us store values and reuse them. Think of them as labeled boxes that hold information.

```python
# Creating variables
starting_balance = 100
grocery_cost = 23.50
side_job_income = 45

print(starting_balance)
```

```python
# Using variables in calculations
final_balance = starting_balance - grocery_cost + side_job_income
print(final_balance)
```

```python
# Variables can be updated
balance = 100
print("Starting balance:", balance)

balance = balance - 23.50
print("After groceries:", balance)

balance = balance + 45
print("After side job:", balance)
```

```python
# Shorthand operators
balance = 100
balance -= 23.50  # Same as: balance = balance - 23.50
balance += 45     # Same as: balance = balance + 45
print(balance)
```

### Variable Naming Rules
- Use lowercase letters and underscores: `my_variable`
- Start with a letter, not a number: `total1` ✓ but `1total` ✗
- No spaces: `total_cost` ✓ but `total cost` ✗
- Be descriptive: `rent` is better than `r`

## Part 3: Build a Budget Calculator (15 minutes)

Let's create a practical budget calculator that computes your monthly finances.

```python
# Budget Calculator - Complete Version

# Income
monthly_salary = 3500
side_income = 400
total_income = monthly_salary + side_income

# Fixed Expenses
rent = 1200
utilities = 150
insurance = 200
car_payment = 350

# Variable Expenses
groceries = 400
dining_out = 200
entertainment = 150
gas = 120

# Calculate totals
total_expenses = rent + utilities + insurance + car_payment + groceries + dining_out + entertainment + gas
remaining_balance = total_income - total_expenses
savings_percentage = (remaining_balance / total_income) * 100

# Display results
print("===== MONTHLY BUDGET REPORT =====")
print(f"Total Income: ${total_income:,.2f}")
print(f"Total Expenses: ${total_expenses:,.2f}")
print(f"Remaining Balance: ${remaining_balance:,.2f}")
print(f"Savings Rate: {savings_percentage:.1f}%")
print("=================================")
```
