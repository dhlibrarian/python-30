## Part 1: Basic Functions (5 minutes)

Functions let you package code for reuse. They're like recipes you can use repeatedly.

```python
# Simple function with no parameters
def greet():
    print("Hello!")

greet()
greet()  # Can call multiple times
```

```python
# Function with parameters
def greet_person(name):
    print(f"Hello, {name}!")

greet_person("Alice")
greet_person("Bob")
```

```python
# Function with multiple parameters
def greet_full(first_name, last_name):
    print(f"Hello, {first_name} {last_name}!")

greet_full("Alice", "Johnson")
```

```python
# Default parameter values
def greet_with_time(name, time="morning"):
    print(f"Good {time}, {name}!")

greet_with_time("Alice")  # Uses default
greet_with_time("Bob", "evening")  # Override default
```

## Part 2: Return Values (5 minutes)

Functions can return values that you can use in your code.

```python
# Function with return value
def add_numbers(a, b):
    result = a + b
    return result

total = add_numbers(5, 3)
print(f"Total: {total}")
```

```python
# Multiple return values
def get_name_parts(full_name):
    parts = full_name.split()
    first = parts[0]
    last = parts[-1]
    return first, last

first_name, last_name = get_name_parts("Alice Johnson")
print(f"First: {first_name}, Last: {last_name}")
```

```python
# Using returned values in calculations
def calculate_area(length, width):
    return length * width

room1_area = calculate_area(10, 12)
room2_area = calculate_area(8, 10)
total_area = room1_area + room2_area

print(f"Total area: {total_area} sq ft")
```

```python
# Boolean returns for validation
def is_valid_email(email):
    return "@" in email and "." in email

print(is_valid_email("user@example.com"))  # True
print(is_valid_email("invalid.com"))  # False
```

## Part 3: Scope and Documentation (5 minutes)

```python
# Variable scope
global_var = "I'm global"

def my_function():
    local_var = "I'm local"
    print(global_var)  # Can access global
    print(local_var)

my_function()
print(global_var)  # Works
# print(local_var)  # Would cause error!
```

```python
# Documenting functions with docstrings
def calculate_bmi(weight_kg, height_m):
    """
    Calculate Body Mass Index (BMI).
    
    Parameters:
        weight_kg (float): Weight in kilograms
        height_m (float): Height in meters
    
    Returns:
        float: BMI value
    """
    bmi = weight_kg / (height_m ** 2)
    return round(bmi, 1)

result = calculate_bmi(70, 1.75)
print(f"BMI: {result}")

# Access documentation
help(calculate_bmi)
```

## Part 4: Unit Converter Library (15 minutes)

Let's build a practical library of conversion functions.

```python
# Unit Converter Library - Complete Version

# Temperature Conversions
def celsius_to_fahrenheit(celsius):
    """Convert Celsius to Fahrenheit."""
    return (celsius * 9/5) + 32

def fahrenheit_to_celsius(fahrenheit):
    """Convert Fahrenheit to Celsius."""
    return (fahrenheit - 32) * 5/9

def celsius_to_kelvin(celsius):
    """Convert Celsius to Kelvin."""
    return celsius + 273.15

# Distance Conversions
def miles_to_kilometers(miles):
    """Convert miles to kilometers."""
    return miles * 1.60934

def kilometers_to_miles(km):
    """Convert kilometers to miles."""
    return km / 1.60934

def feet_to_meters(feet):
    """Convert feet to meters."""
    return feet * 0.3048

def meters_to_feet(meters):
    """Convert meters to feet."""
    return meters / 0.3048

# Weight Conversions
def pounds_to_kilograms(pounds):
    """Convert pounds to kilograms."""
    return pounds * 0.453592

def kilograms_to_pounds(kg):
    """Convert kilograms to pounds."""
    return kg / 0.453592

# Volume Conversions
def gallons_to_liters(gallons):
    """Convert gallons to liters."""
    return gallons * 3.78541

def liters_to_gallons(liters):
    """Convert liters to gallons."""
    return liters / 3.78541

# Helper function for formatted output
def convert_and_print(value, from_unit, to_unit, conversion_func):
    """
    Convert value and print formatted result.
    """
    result = conversion_func(value)
    print(f"{value} {from_unit} = {result:.2f} {to_unit}")
    return result

# Test the library
print("===== UNIT CONVERTER LIBRARY =====")
print("\nTemperature:")
convert_and_print(100, "°C", "°F", celsius_to_fahrenheit)
convert_and_print(32, "°F", "°C", fahrenheit_to_celsius)

print("\nDistance:")
convert_and_print(10, "miles", "km", miles_to_kilometers)
convert_and_print(5, "feet", "meters", feet_to_meters)

print("\nWeight:")
convert_and_print(150, "lbs", "kg", pounds_to_kilograms)

print("\nVolume:")
convert_and_print(5, "gallons", "liters", gallons_to_liters)
print("=" * 35)
```

### Interactive Converter

Let's create an interactive version where you can easily convert values.

```python
def universal_converter(value, from_unit, to_unit):
    """
    Universal converter that handles multiple unit types.
    
    Supported conversions:
    Temperature: C, F, K
    Distance: miles, km, feet, meters
    Weight: lbs, kg
    Volume: gallons, liters
    """
    conversions = {
        ("C", "F"): celsius_to_fahrenheit,
        ("F", "C"): fahrenheit_to_celsius,
        ("C", "K"): celsius_to_kelvin,
        ("miles", "km"): miles_to_kilometers,
        ("km", "miles"): kilometers_to_miles,
        ("feet", "meters"): feet_to_meters,
        ("meters", "feet"): meters_to_feet,
        ("lbs", "kg"): pounds_to_kilograms,
        ("kg", "lbs"): kilograms_to_pounds,
        ("gallons", "liters"): gallons_to_liters,
        ("liters", "gallons"): liters_to_gallons
    }
    
    key = (from_unit, to_unit)
    if key in conversions:
        result = conversions[key](value)
        print(f"{value} {from_unit} = {result:.2f} {to_unit}")
        return result
    else:
        print(f"Conversion from {from_unit} to {to_unit} not supported")
        return None

# Test it
universal_converter(100, "C", "F")
universal_converter(5, "miles", "km")
universal_converter(10, "gallons", "liters")
```

### Enhanced Version: Batch Converter

```python
def batch_convert(values, from_unit, to_unit):
    """
    Convert multiple values at once.
    """
    print(f"\nConverting {from_unit} to {to_unit}:")
    print("-" * 40)
    
    results = []
    for value in values:
        result = universal_converter(value, from_unit, to_unit)
        if result is not None:
            results.append(result)
    
    return results

# Convert multiple temperatures
celsius_temps = [0, 20, 37, 100]
fahrenheit_temps = batch_convert(celsius_temps, "C", "F")
```
