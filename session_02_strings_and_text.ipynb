## Part 1: Introduction to Strings (5 minutes)

Strings are sequences of characters (text). You can create them with single or double quotes.

```python
# Creating strings
greeting = "Hello"
name = 'Python'

print(greeting)
print(name)
```

```python
# String repetition
laugh = "ha" * 5
print(laugh)
```

## Part 2: String Methods (10 minutes)

Python strings have built-in methods for common text operations.

```python
# Changing case
text = "Python is Amazing"

print(text.upper())      # PYTHON IS AMAZING
print(text.lower())      # python is amazing
print(text.title())      # Python Is Amazing
```

```python
# Removing whitespace
messy = "  too much space  "
print(f"'{messy}'")
print(f"'{messy.strip()}'")
```

```python
# Replacing text
sentence = "I love cats"
new_sentence = sentence.replace("cats", "dogs")
print(new_sentence)
```

```python
# Splitting strings into lists
email = "john.doe@example.com"
parts = email.split("@")
print(parts)
print(f"Username: {parts[0]}")
print(f"Domain: {parts[1]}")
```

```python
# Checking string contents
filename = "document.pdf"
print(filename.startswith("doc"))  # True
print(filename.endswith(".pdf"))   # True
print("pdf" in filename)           # True
```

```python
# Getting string length
message = "Hello, World!"
print(f"The message has {len(message)} characters")
```

## Part 3: F-Strings for Formatting (5 minutes)

F-strings (formatted string literals) make it easy to insert variables into text.

```python
# Old way (don't use this)
name = "Alice"
age = 30
print("Hello " + name + ", you are " + str(age) + " years old")
```

```python
# Better way with f-strings (use this!)
name = "Alice"
age = 30
print(f"Hello {name}, you are {age} years old")
```

```python
# F-strings can include expressions
price = 19.99
quantity = 3
print(f"Total cost: ${price * quantity:.2f}")
```

```python
# F-strings can call methods
name = "python"
print(f"I love {name.upper()}!")
```

## Part 4: Email Template Generator (10 minutes)

Let's build a practical tool that generates personalized emails!

```python
# Email Template Generator - Example

# Recipient information
recipient_name = "Dr. Smith"
recipient_title = "Professor"
recipient_department = "History"

# Sender information
sender_name = "Joshua"
sender_title = "Digital Scholarship Librarian"

# Email details
subject = "Workshop on Digital Research Tools"
event_date = "March 15, 2024"
event_time = "2:00 PM"
location = "Library Computer Lab"

# Generate the email
email = f"""
Subject: {subject}

Dear {recipient_title} {recipient_name},

I hope this message finds you well. I am writing to invite you and your students in the {recipient_department} department to an upcoming workshop on digital research tools.

Event Details:
- Date: {event_date}
- Time: {event_time}
- Location: {location}

This workshop will cover practical tools for digital humanities research, including citation management and text analysis software. I believe this would be valuable for your students' research projects.

Please let me know if you or your students would like to attend. I'm happy to adjust the timing if needed.

Best regards,
{sender_name}
{sender_title}
"""

print(email)
```
