## Part 1: Introduction to Lists (5 minutes)

Lists are ordered collections that can hold multiple items. They're one of Python's most versatile data structures.

```python
# Creating lists
fruits = ["apple", "banana", "cherry"]
numbers = [1, 2, 3, 4, 5]
mixed = ["text", 42, 3.14, True]

print(fruits)
print(numbers)
print(mixed)
```

## Part 2: Accessing List Items (5 minutes)

Lists are indexed starting at 0. You can access items by their position.

```python
# Indexing (positions start at 0)
colors = ["red", "green", "blue", "yellow", "purple"]

print(colors[0])   # First item
print(colors[2])   # Third item
print(colors[-1])  # Last item
print(colors[-2])  # Second to last
```

```python
# Slicing (getting multiple items)
colors = ["red", "green", "blue", "yellow", "purple"]

print(colors[1:3])    # Items at index 1 and 2
print(colors[:3])     # First 3 items
print(colors[2:])     # From index 2 to end
print(colors[-2:])    # Last 2 items
```

```python
# Checking if item exists
if "blue" in colors:
    print("Blue is in the list!")

if "orange" not in colors:
    print("Orange is not in the list")
```

## Part 3: Modifying Lists (10 minutes)

Lists are mutable, meaning you can change them after creation.

```python
# Adding items
shopping = ["milk", "bread"]
print("Original:", shopping)

shopping.append("eggs")  # Add to end
print("After append:", shopping)

shopping.insert(1, "butter")  # Insert at position 1
print("After insert:", shopping)
```

```python
# Removing items
tasks = ["email", "meeting", "lunch", "report"]
print("Original:", tasks)

tasks.remove("lunch")  # Remove by value
print("After remove:", tasks)

completed = tasks.pop()  # Remove and return last item
print(f"Completed: {completed}")
print("Remaining:", tasks)
```

```python
# Changing items
scores = [85, 90, 78, 92]
print("Original:", scores)

scores[2] = 88  # Update third score
print("Updated:", scores)
```

```python
# Sorting lists
numbers = [5, 2, 8, 1, 9]
numbers.sort()  # Sort in place
print("Sorted:", numbers)

numbers.sort(reverse=True)  # Sort descending
print("Reverse sorted:", numbers)
```

```python
# Combining lists
list1 = [1, 2, 3]
list2 = [4, 5, 6]
combined = list1 + list2
print(combined)
```

```python
# Useful list methods
numbers = [1, 2, 3, 2, 4, 2, 5]

print(f"Count of 2: {numbers.count(2)}")
print(f"Index of first 4: {numbers.index(4)}")
print(f"Sum: {sum(numbers)}")
print(f"Max: {max(numbers)}")
print(f"Min: {min(numbers)}")
```

## Part 4: Reading List Manager (10 minutes)

Let's build a practical reading list manager with add, remove, and display functions.

```python
# Reading List Manager - Complete Version

# Initialize empty reading list
reading_list = []

# Add books to the list
def add_book(title, author):
    book = f"{title} by {author}"
    reading_list.append(book)
    print(f"✓ Added: {book}")

# Remove a book
def remove_book(title):
    for book in reading_list:
        if title.lower() in book.lower():
            reading_list.remove(book)
            print(f"✓ Removed: {book}")
            return
    print(f"✗ Book not found: {title}")

# Display all books
def display_list():
    if not reading_list:
        print("Your reading list is empty!")
        return
    
    print("\n=== MY READING LIST ===")
    for i, book in enumerate(reading_list, 1):
        print(f"{i}. {book}")
    print(f"\nTotal books: {len(reading_list)}")
    print("=" * 25)

# Test the manager
add_book("1984", "George Orwell")
add_book("To Kill a Mockingbird", "Harper Lee")
add_book("The Great Gatsby", "F. Scott Fitzgerald")
display_list()

remove_book("1984")
display_list()
```

### Enhanced Version: Reading List with Status

Let's enhance our reading list to track reading status.

```python
# Enhanced Reading List with dictionaries
reading_list_enhanced = []

def add_book_enhanced(title, author, status="to-read"):
    book = {
        "title": title,
        "author": author,
        "status": status  # to-read, reading, finished
    }
    reading_list_enhanced.append(book)
    print(f"✓ Added: {title}")

def update_status(title, new_status):
    for book in reading_list_enhanced:
        if title.lower() in book["title"].lower():
            book["status"] = new_status
            print(f"✓ Updated {book['title']} to: {new_status}")
            return
    print(f"✗ Book not found: {title}")

def display_enhanced():
    if not reading_list_enhanced:
        print("Your reading list is empty!")
        return
    
    print("\n=== MY READING LIST ===")
    for book in reading_list_enhanced:
        status_icon = {"to-read": "📚", "reading": "📖", "finished": "✅"}
        icon = status_icon.get(book["status"], "")
        print(f"{icon} {book['title']} by {book['author']} [{book['status']}]")
    print("=" * 40)

# Test enhanced version
add_book_enhanced("Dune", "Frank Herbert")
add_book_enhanced("Neuromancer", "William Gibson", "reading")
add_book_enhanced("Foundation", "Isaac Asimov", "finished")
display_enhanced()

update_status("Dune", "reading")
display_enhanced()
```
