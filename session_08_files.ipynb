## Part 1: Reading Text Files (5 minutes)

```python
# First, let's create a sample file
with open('sample.txt', 'w') as file:
    file.write("Hello, World!\n")
    file.write("This is a sample file.\n")
    file.write("Python makes file handling easy!")

print("File created successfully")
```

```python
# Reading entire file
with open('sample.txt', 'r') as file:
    content = file.read()
    print(content)
```

```python
# Reading line by line
with open('sample.txt', 'r') as file:
    for line in file:
        print(line.strip())  # .strip() removes newline
```

```python
# Reading all lines into a list
with open('sample.txt', 'r') as file:
    lines = file.readlines()
    print(f"File has {len(lines)} lines")
    print(lines)
```

## Part 2: Writing to Files (5 minutes)

```python
# Writing to a file (overwrites existing content)
with open('output.txt', 'w') as file:
    file.write("This is line 1\n")
    file.write("This is line 2\n")

print("File written successfully")
```

```python
# Appending to a file (adds to existing content)
with open('output.txt', 'a') as file:
    file.write("This is line 3\n")
    file.write("This is line 4\n")

# Read it back
with open('output.txt', 'r') as file:
    print(file.read())
```

```python
# Writing multiple lines at once
lines = [
    "Python is great\n",
    "For data analysis\n",
    "And automation\n"
]

with open('multi_lines.txt', 'w') as file:
    file.writelines(lines)

print("Multiple lines written")
```

## Part 3: Working with CSV Files (5 minutes)

```python
import csv

# Writing CSV
data = [
    ['Name', 'Age', 'City'],
    ['Alice', 30, 'New York'],
    ['Bob', 25, 'Los Angeles'],
    ['Charlie', 35, 'Chicago']
]

with open('people.csv', 'w', newline='') as file:
    writer = csv.writer(file)
    writer.writerows(data)

print("CSV file created")
```

```python
# Reading CSV
with open('people.csv', 'r') as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

```python
# Working with CSV as dictionaries
with open('people.csv', 'r') as file:
    reader = csv.DictReader(file)
    for row in reader:
        print(f"{row['Name']} is {row['Age']} years old")
```

## Part 4: Journal System (15 minutes)

Let's build a practical journaling system that saves and retrieves entries.

```python
# Journal System - Complete Version

from datetime import datetime
import os

JOURNAL_FILE = 'my_journal.txt'

def add_entry(entry_text):
    """
    Add a new journal entry with timestamp.
    """
    timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    
    with open(JOURNAL_FILE, 'a') as file:
        file.write(f"\n{'='*60}\n")
        file.write(f"Date: {timestamp}\n")
        file.write(f"{'-'*60}\n")
        file.write(f"{entry_text}\n")
    
    print(f"✓ Journal entry added at {timestamp}")

def view_all_entries():
    """
    Display all journal entries.
    """
    if not os.path.exists(JOURNAL_FILE):
        print("No journal entries yet. Start writing!")
        return
    
    print("\n===== MY JOURNAL =====")
    with open(JOURNAL_FILE, 'r') as file:
        print(file.read())
    print("=" * 22)

def count_entries():
    """
    Count the number of entries.
    """
    if not os.path.exists(JOURNAL_FILE):
        return 0
    
    with open(JOURNAL_FILE, 'r') as file:
        content = file.read()
        return content.count('Date:')

def search_entries(keyword):
    """
    Search for entries containing a keyword.
    """
    if not os.path.exists(JOURNAL_FILE):
        print("No journal entries to search.")
        return
    
    with open(JOURNAL_FILE, 'r') as file:
        content = file.read()
    
    entries = content.split('='*60)
    matches = [entry for entry in entries if keyword.lower() in entry.lower()]
    
    if matches:
        print(f"\nFound {len(matches)} entry/entries containing '{keyword}':")
        for match in matches:
            if match.strip():
                print('='*60)
                print(match.strip())
    else:
        print(f"No entries found containing '{keyword}'")

def get_recent_entries(n=5):
    """
    Get the n most recent entries.
    """
    if not os.path.exists(JOURNAL_FILE):
        print("No journal entries yet.")
        return
    
    with open(JOURNAL_FILE, 'r') as file:
        content = file.read()
    
    entries = [e for e in content.split('='*60) if e.strip()]
    recent = entries[-n:] if len(entries) >= n else entries
    
    print(f"\n===== RECENT {len(recent)} ENTRIES =====")
    for entry in recent:
        print('='*60)
        print(entry.strip())
    print("=" * 30)

# Test the journal system
add_entry("Today was a great day! I learned about file handling in Python.")
add_entry("Working on my journaling app. It's coming together nicely!")
add_entry("Feeling productive. Completed three sessions today.")

print(f"\nTotal entries: {count_entries()}")
view_all_entries()
```

```python
# Search functionality
search_entries("Python")
```

```python
# View recent entries
get_recent_entries(2)
```

### Enhanced Version: Tagged Entries

```python
import csv

TAGGED_JOURNAL = 'tagged_journal.csv'

def add_tagged_entry(entry_text, tags=None):
    """
    Add entry with tags to CSV journal.
    """
    timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    tags_str = ','.join(tags) if tags else ''
    
    # Create file with headers if it doesn't exist
    file_exists = os.path.exists(TAGGED_JOURNAL)
    
    with open(TAGGED_JOURNAL, 'a', newline='') as file:
        writer = csv.writer(file)
        if not file_exists:
            writer.writerow(['Timestamp', 'Entry', 'Tags'])
        writer.writerow([timestamp, entry_text, tags_str])
    
    print(f"✓ Tagged entry added with tags: {tags_str}")

def view_by_tag(tag):
    """
    View all entries with a specific tag.
    """
    if not os.path.exists(TAGGED_JOURNAL):
        print("No tagged entries yet.")
        return
    
    with open(TAGGED_JOURNAL, 'r') as file:
        reader = csv.DictReader(file)
        matches = [row for row in reader if tag in row['Tags']]
    
    if matches:
        print(f"\nEntries tagged with '{tag}':")
        for entry in matches:
            print(f"\n{entry['Timestamp']}")
            print(f"{entry['Entry']}")
            print(f"Tags: {entry['Tags']}")
    else:
        print(f"No entries found with tag '{tag}'")

# Test tagged journal
add_tagged_entry("Great Python session today!", ["python", "learning"])
add_tagged_entry("Need to review file handling.", ["python", "todo"])
add_tagged_entry("Finished my project!", ["achievement", "project"])

view_by_tag("python")
```
