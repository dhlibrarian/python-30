```python
import pandas as pd

# Create messy dataset
data = {
    'Name': ['Alice', 'Bob', None, 'Diana', 'Eve'],
    'Age': [25, 30, 35, None, 28],
    'City': ['New York', 'los angeles', 'CHICAGO', 'Houston', 'new york'],
    'Salary': [70000, None, 90000, 75000, 80000],
    'Department': ['Sales', 'Sales', 'Marketing', 'Sales', 'Marketing']
}

df = pd.DataFrame(data)
print('Original Data:')
print(df)
```

## Part 1: Handling Missing Values (10 minutes)

Real-world data is rarely clean. Missing values, inconsistent capitalization, and unlabeled categories are the norm — this session covers the pandas tools for handling them.

```python
# Check for missing values
print('Missing values:')
print(df.isnull().sum())

# Fill missing values
# Note: df['col'] = df['col'].fillna(...) is the recommended form -
# df['col'].fillna(..., inplace=True) is deprecated because it can silently
# fail to update the original DataFrame.
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Salary'] = df['Salary'].fillna(df['Salary'].median())
df['Name'] = df['Name'].fillna('Unknown')

print('\nAfter filling:')
print(df)
```

## Part 2: Cleaning Text Data (5 minutes)

```python
# Standardize text
df['City'] = df['City'].str.title()
print('Cleaned cities:')
print(df['City'])
```

## Part 3: Creating New Columns & Grouping (5 minutes)

```python
# Create salary category
df['Salary_Category'] = pd.cut(df['Salary'], 
                                bins=[0, 75000, 85000, 100000],
                                labels=['Low', 'Medium', 'High'])
print(df[['Name', 'Salary', 'Salary_Category']])
```

### Grouping and Aggregation

```python
# Group by department
dept_stats = df.groupby('Department').agg({
    'Salary': ['mean', 'min', 'max'],
    'Age': 'mean'
}).round(2)

print('Department Statistics:')
print(dept_stats)
```

## Part 4: Practical Project - Clean Sales Data (10 minutes)

Let's put everything from this session together on a small, messy sales dataset.

```python
# Create sample sales data
sales_data = {
    'Date': ['2024-01-01', '2024-01-02', '2024-01-03', None, '2024-01-05'],
    'Product': ['laptop', 'MOUSE', 'Keyboard', 'laptop', 'mouse'],
    'Quantity': [2, None, 5, 1, 3],
    'Price': [1000, 25, 75, 1000, 25],
    'Region': ['North', 'south', 'EAST', 'North', 'South']
}

sales = pd.DataFrame(sales_data)

# Clean the data
sales['Date'] = pd.to_datetime(sales['Date'])
sales['Product'] = sales['Product'].str.title()
sales['Region'] = sales['Region'].str.title()
sales['Quantity'] = sales['Quantity'].fillna(1)
sales['Total'] = sales['Quantity'] * sales['Price']

print('Cleaned Sales Data:')
print(sales)

# Analysis
print('\nSummary by Product:')
print(sales.groupby('Product')['Total'].sum().sort_values(ascending=False))
```
