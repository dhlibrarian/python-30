# Session 11: Data Visualization with Matplotlib

Topics: Line plots, bar charts, scatter plots, customization

**Duration:** 30 minutes

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Enable inline plotting
%matplotlib inline
```

## Basic Line Plot

```python
# Simple line plot
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
sales = [15000, 18000, 22000, 20000, 25000, 30000]

plt.figure(figsize=(10, 6))
plt.plot(months, sales, marker='o', linewidth=2, color='blue')
plt.title('Monthly Sales', fontsize=16)
plt.xlabel('Month', fontsize=12)
plt.ylabel('Sales ($)', fontsize=12)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## Bar Chart

```python
# Bar chart
products = ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Headphones']
quantities = [45, 120, 85, 30, 95]

plt.figure(figsize=(10, 6))
plt.bar(products, quantities, color=['#FF6B6B', '#4ECDC4', '#45B7D1', '#FFA07A', '#98D8C8'])
plt.title('Product Sales by Quantity', fontsize=16)
plt.xlabel('Product', fontsize=12)
plt.ylabel('Quantity Sold', fontsize=12)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Scatter Plot

```python
# Scatter plot
study_hours = [1, 2, 3, 4, 5, 6, 7, 8]
test_scores = [65, 70, 75, 78, 82, 85, 88, 90]

plt.figure(figsize=(10, 6))
plt.scatter(study_hours, test_scores, s=100, alpha=0.6, c='green')
plt.title('Study Hours vs Test Scores', fontsize=16)
plt.xlabel('Hours Studied', fontsize=12)
plt.ylabel('Test Score', fontsize=12)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## Multiple Plots

```python
# Subplots
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# Plot 1: Line
axes[0, 0].plot(months, sales, marker='o')
axes[0, 0].set_title('Sales Trend')
axes[0, 0].grid(True, alpha=0.3)

# Plot 2: Bar
axes[0, 1].bar(products, quantities, color='skyblue')
axes[0, 1].set_title('Product Quantities')
axes[0, 1].tick_params(axis='x', rotation=45)

# Plot 3: Scatter
axes[1, 0].scatter(study_hours, test_scores, c='red', alpha=0.6)
axes[1, 0].set_title('Study vs Scores')
axes[1, 0].grid(True, alpha=0.3)

# Plot 4: Histogram
data = np.random.normal(100, 15, 1000)
axes[1, 1].hist(data, bins=30, color='purple', alpha=0.7)
axes[1, 1].set_title('Distribution')

plt.tight_layout()
plt.show()
```

## Practical Project: Personal Finance Dashboard

```python
# Create personal finance data
expense_categories = ['Rent', 'Food', 'Transport', 'Entertainment', 'Utilities', 'Savings']
monthly_amounts = [1200, 400, 200, 150, 100, 450]

income = 2500
total_expenses = sum(monthly_amounts)
savings_rate = (monthly_amounts[-1] / income) * 100

# Create visualization dashboard
fig = plt.figure(figsize=(15, 10))

# Pie chart
ax1 = plt.subplot(2, 2, 1)
colors = ['#FF6B6B', '#4ECDC4', '#45B7D1', '#FFA07A', '#98D8C8', '#90EE90']
ax1.pie(monthly_amounts, labels=expense_categories, autopct='%1.1f%%', colors=colors)
ax1.set_title('Expense Breakdown', fontsize=14, fontweight='bold')

# Bar chart
ax2 = plt.subplot(2, 2, 2)
ax2.bar(expense_categories, monthly_amounts, color=colors)
ax2.set_title('Monthly Expenses by Category', fontsize=14, fontweight='bold')
ax2.set_ylabel('Amount ($)')
ax2.tick_params(axis='x', rotation=45)

# Income vs Expenses
ax3 = plt.subplot(2, 2, 3)
categories = ['Income', 'Expenses', 'Savings']
values = [income, total_expenses, monthly_amounts[-1]]
ax3.bar(categories, values, color=['green', 'red', 'blue'])
ax3.set_title('Income vs Expenses', fontsize=14, fontweight='bold')
ax3.set_ylabel('Amount ($)')

# Summary text
ax4 = plt.subplot(2, 2, 4)
ax4.axis('off')
summary_text = f"""
MONTHLY FINANCIAL SUMMARY

Total Income: ${income:,.2f}
Total Expenses: ${total_expenses:,.2f}
Net Savings: ${income - total_expenses:,.2f}

Savings Rate: {savings_rate:.1f}%

Largest Expense: {expense_categories[monthly_amounts.index(max(monthly_amounts))]}
Amount: ${max(monthly_amounts):,.2f}
"""
ax4.text(0.1, 0.5, summary_text, fontsize=12, family='monospace',
         verticalalignment='center')

plt.tight_layout()
plt.savefig('finance_dashboard.png', dpi=300, bbox_inches='tight')
plt.show()

print('Dashboard saved as finance_dashboard.png')
```
