## Part 1: Basic If Statements (5 minutes)

Conditionals let your program make decisions based on conditions.

```python
# Simple if statement
temperature = 75

if temperature > 70:
    print("It's warm outside!")
```

```python
# If-else statement
temperature = 65

if temperature > 70:
    print("It's warm outside!")
else:
    print("It's cool outside!")
```

```python
# If-elif-else for multiple conditions
temperature = 85

if temperature > 80:
    print("It's hot! 🔥")
elif temperature > 70:
    print("It's warm! ☀️")
elif temperature > 50:
    print("It's cool! 🍂")
else:
    print("It's cold! ❄️")
```

## Part 2: Comparison Operators (5 minutes)

These operators compare values and return True or False.

```python
# Comparison operators
x = 10
y = 5

print(f"x == y: {x == y}")   # Equal to
print(f"x != y: {x != y}")   # Not equal to
print(f"x > y: {x > y}")     # Greater than
print(f"x < y: {x < y}")     # Less than
print(f"x >= y: {x >= y}")   # Greater than or equal
print(f"x <= y: {x <= y}")   # Less than or equal
```

```python
# String comparisons
name = "Alice"

if name == "Alice":
    print("Hello, Alice!")

# Case-sensitive
if name.lower() == "alice":
    print("This handles uppercase/lowercase")
```

```python
# Checking membership
fruits = ["apple", "banana", "cherry"]

if "banana" in fruits:
    print("We have bananas!")

if "orange" not in fruits:
    print("No oranges available")
```

## Part 3: Logical Operators (5 minutes)

Combine multiple conditions using `and`, `or`, and `not`.

```python
# AND operator - both must be True
temperature = 75
is_sunny = True

if temperature > 70 and is_sunny:
    print("Perfect day for a picnic!")
```

```python
# OR operator - at least one must be True
day = "Saturday"

if day == "Saturday" or day == "Sunday":
    print("It's the weekend!")
```

```python
# NOT operator - reverses the condition
is_raining = False

if not is_raining:
    print("No need for an umbrella")
```

```python
# Combining multiple conditions
age = 25
has_license = True
has_car = False

if age >= 18 and has_license and has_car:
    print("You can drive your own car")
elif age >= 18 and has_license:
    print("You can drive a rental car")
else:
    print("You cannot drive")
```

```python
# Checking ranges
score = 85

if 80 <= score <= 89:  # Pythonic way!
    print("Grade: B")

# Same as:
if score >= 80 and score <= 89:
    print("Grade: B")
```

## Part 4: Decision-Making Tool (15 minutes)

Let's build a practical tool that helps decide whether to bike or drive based on conditions.

**A quick preview:** this tool is wrapped in a *function*, `def should_bike_or_drive(...):`, so we can reuse the same decision logic for as many scenarios as we like just by calling `should_bike_or_drive(weather=..., temperature=..., ...)` again with new values — that's exactly what the cells below do. Functions are covered in depth in Session 7; for now, treat everything indented under `def should_bike_or_drive(...):` as the tool's recipe, and each call below as "run that recipe with these inputs."

```python
# Bike or Drive Decision Tool - Complete Version

def should_bike_or_drive(weather, temperature, distance_miles, time_available_mins, fitness_level):
    """
    Recommends biking or driving based on various factors.
    
    Parameters:
    - weather: "sunny", "cloudy", "rainy", "snowy"
    - temperature: temperature in Fahrenheit
    - distance_miles: distance to destination
    - time_available_mins: available time in minutes
    - fitness_level: "low", "medium", "high"
    """
    
    reasons = []
    score = 0  # Positive = bike, Negative = drive
    
    # Check weather conditions
    if weather == "sunny" or weather == "cloudy":
        score += 2
        reasons.append("✓ Weather is good for biking")
    elif weather == "rainy":
        score -= 3
        reasons.append("✗ Rainy weather favors driving")
    elif weather == "snowy":
        score -= 5
        reasons.append("✗ Snow makes biking dangerous")
    
    # Check temperature
    if 60 <= temperature <= 80:
        score += 2
        reasons.append("✓ Temperature is comfortable for biking")
    elif temperature > 90 or temperature < 40:
        score -= 2
        reasons.append("✗ Temperature is uncomfortable for biking")
    
    # Check distance
    if distance_miles <= 3:
        score += 3
        reasons.append("✓ Distance is bikeable")
    elif distance_miles <= 6:
        score += 1
        reasons.append("≈ Distance is moderate for biking")
    else:
        score -= 2
        reasons.append("✗ Distance is long for biking")
    
    # Check time (assume 10 mph biking, 30 mph driving)
    bike_time = (distance_miles / 10) * 60
    drive_time = (distance_miles / 30) * 60
    
    if bike_time <= time_available_mins:
        score += 1
        reasons.append(f"✓ You have enough time ({time_available_mins} mins available, {bike_time:.0f} mins needed)")
    else:
        score -= 3
        reasons.append(f"✗ Not enough time for biking ({time_available_mins} mins available, {bike_time:.0f} mins needed)")
    
    # Check fitness level
    if fitness_level == "high":
        score += 2
        reasons.append("✓ High fitness level supports biking")
    elif fitness_level == "low":
        score -= 1
        reasons.append("≈ Low fitness level may make biking harder")
    
    # Make recommendation
    print("\n===== BIKE OR DRIVE RECOMMENDATION =====")
    print(f"Distance: {distance_miles} miles")
    print(f"Weather: {weather.title()}, {temperature}°F")
    print(f"Time available: {time_available_mins} minutes")
    print(f"Fitness level: {fitness_level.title()}")
    print("\nAnalysis:")
    for reason in reasons:
        print(f"  {reason}")
    
    print(f"\nDecision Score: {score}")
    if score >= 3:
        print("\n🚴 RECOMMENDATION: BIKE")
        print(f"Estimated biking time: {bike_time:.0f} minutes")
    elif score <= -3:
        print("\n🚗 RECOMMENDATION: DRIVE")
        print(f"Estimated driving time: {drive_time:.0f} minutes")
    else:
        print("\n🤷 RECOMMENDATION: EITHER WORKS")
        print(f"Biking: {bike_time:.0f} mins | Driving: {drive_time:.0f} mins")
    print("=" * 40)

# Test the tool
should_bike_or_drive(
    weather="sunny",
    temperature=72,
    distance_miles=4,
    time_available_mins=40,
    fitness_level="medium"
)
```

### Simplified Version: Quick Decision Maker

A simpler version focusing on just the key factors.

```python
# Quick decision maker
def quick_decision(weather, distance):
    print(f"\nConditions: {weather} weather, {distance} miles")
    
    if weather == "rainy" or weather == "snowy":
        print("Decision: 🚗 DRIVE (bad weather)")
    elif distance > 5:
        print("Decision: 🚗 DRIVE (too far)")
    elif distance <= 2:
        print("Decision: 🚴 BIKE (short distance, good weather)")
    else:
        print("Decision: 🚴 BIKE (moderate distance, good weather)")

# Test it
quick_decision("sunny", 3)
quick_decision("rainy", 1)
quick_decision("sunny", 8)
```
