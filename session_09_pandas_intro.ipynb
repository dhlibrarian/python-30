## Part 1: Introduction to Pandas (5 minutes)

Pandas is Python's primary library for data analysis. It provides DataFrames, which are like super-powered spreadsheets.

```python
import pandas as pd
import numpy as np

# Creating a simple DataFrame
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'Age': [25, 30, 35, 28],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Houston'],
    'Salary': [70000, 80000, 90000, 75000]
}

df = pd.DataFrame(data)
print(df)
```

```python
# Viewing DataFrame info
print("Shape:", df.shape)  # (rows, columns)
print("\nColumns:", df.columns.tolist())
print("\nData types:")
print(df.dtypes)
```

```python
# Quick statistics
df.describe()
```

### A Word About NumPy

You may have noticed `import numpy as np` at the top. Pandas is actually built on top of NumPy, and NumPy arrays are what let pandas do fast math on entire columns at once (instead of looping through values one by one). You won't need to write much NumPy code directly today, but here's a quick look at what it's doing behind the scenes.

```python
# A pandas column (Series) is backed by a NumPy array
salary_array = np.array(df['Salary'])
print("As a NumPy array:", salary_array)
print("Type:", type(salary_array))

# NumPy functions work directly on the whole array at once - no loop needed
print(f"\nAverage salary (np.mean): {np.mean(salary_array):,.2f}")
print(f"Salary standard deviation (np.std): {np.std(salary_array):,.2f}")
```

## Part 3: Filtering and Selection (5 minutes)

```python
# Filtering rows based on condition
high_earners = df[df['Salary'] > 75000]
print("High earners:")
print(high_earners)
```

```python
# Multiple conditions
young_high_earners = df[(df['Age'] < 30) & (df['Salary'] > 70000)]
print("Young high earners:")
print(young_high_earners)
```

```python
# Filtering with string methods
ny_people = df[df['City'].str.contains('New')]
print(ny_people)
```

```python
# Sorting
df_sorted = df.sort_values('Salary', ascending=False)
print("Sorted by salary (high to low):")
print(df_sorted)
```

## Part 4: Reading and Analyzing Real Data (15 minutes)

Let's create and analyze a book ratings dataset.

```python
# Create sample book dataset
books_data = {
    'Title': [
        '1984', 'To Kill a Mockingbird', 'The Great Gatsby',
        'Pride and Prejudice', 'The Catcher in the Rye',
        'Animal Farm', 'Brave New World', 'Lord of the Flies',
        'The Hobbit', 'Fahrenheit 451'
    ],
    'Author': [
        'George Orwell', 'Harper Lee', 'F. Scott Fitzgerald',
        'Jane Austen', 'J.D. Salinger',
        'George Orwell', 'Aldous Huxley', 'William Golding',
        'J.R.R. Tolkien', 'Ray Bradbury'
    ],
    'Year': [1949, 1960, 1925, 1813, 1951, 1945, 1932, 1954, 1937, 1953],
    'Pages': [328, 281, 180, 432, 234, 112, 268, 224, 310, 249],
    'Rating': [4.7, 4.8, 4.4, 4.6, 4.0, 4.5, 4.3, 4.2, 4.8, 4.5],
    'Genre': [
        'Dystopian', 'Classic', 'Classic', 'Romance', 'Coming-of-age',
        'Dystopian', 'Dystopian', 'Allegory', 'Fantasy', 'Dystopian'
    ]
}

books = pd.DataFrame(books_data)

# Save to CSV
books.to_csv('books.csv', index=False)
print("Dataset created and saved to books.csv")
print("\nFirst few rows:")
books.head()
```

```python
# Reading CSV file
books_df = pd.read_csv('books.csv')
print("Loaded dataset:")
print(f"Shape: {books_df.shape}")
print(f"Columns: {books_df.columns.tolist()}")
```

```python
# Basic analysis
print("===== BOOK DATASET ANALYSIS =====")
print(f"\nTotal books: {len(books_df)}")
print(f"\nAverage rating: {books_df['Rating'].mean():.2f}")
print(f"Highest rating: {books_df['Rating'].max()}")
print(f"Lowest rating: {books_df['Rating'].min()}")
print(f"\nAverage pages: {books_df['Pages'].mean():.0f}")
print(f"\nOldest book: {books_df['Year'].min()}")
print(f"Newest book: {books_df['Year'].max()}")
```

```python
# Top rated books
print("\nTop 3 Rated Books:")
top_books = books_df.nlargest(3, 'Rating')[['Title', 'Author', 'Rating']]
print(top_books)
```

```python
# Books by genre
print("\nBooks by Genre:")
genre_counts = books_df['Genre'].value_counts()
print(genre_counts)
```

```python
# Average rating by genre
print("\nAverage Rating by Genre:")
genre_ratings = books_df.groupby('Genre')['Rating'].mean().sort_values(ascending=False)
for genre, rating in genre_ratings.items():
    print(f"{genre}: {rating:.2f}")
```

```python
# Find specific books
print("\nGeorge Orwell's books:")
orwell_books = books_df[books_df['Author'] == 'George Orwell']
print(orwell_books[['Title', 'Year', 'Rating']])
```

```python
# High-rated, longer books
print("\nHigh-rated books over 300 pages:")
long_good_books = books_df[(books_df['Rating'] >= 4.5) & (books_df['Pages'] > 300)]
print(long_good_books[['Title', 'Pages', 'Rating']])
```

### Complete Analysis Report

```python
def generate_book_report(df):
    """
    Generate comprehensive analysis report.
    """
    print("\n" + "="*50)
    print("COMPREHENSIVE BOOK ANALYSIS REPORT")
    print("="*50)
    
    print(f"\n📚 Dataset Overview:")
    print(f"   Total Books: {len(df)}")
    print(f"   Unique Authors: {df['Author'].nunique()}")
    print(f"   Genres: {df['Genre'].nunique()}")
    print(f"   Year Range: {df['Year'].min()} - {df['Year'].max()}")
    
    print(f"\n⭐ Rating Statistics:")
    print(f"   Average: {df['Rating'].mean():.2f}")
    print(f"   Median: {df['Rating'].median():.2f}")
    print(f"   Range: {df['Rating'].min():.1f} - {df['Rating'].max():.1f}")
    
    print(f"\n📖 Page Statistics:")
    print(f"   Average: {df['Pages'].mean():.0f}")
    print(f"   Shortest: {df['Pages'].min()} pages")
    print(f"   Longest: {df['Pages'].max()} pages")
    
    print(f"\n🏆 Top 3 Books:")
    top_3 = df.nlargest(3, 'Rating')
    for idx, (_, book) in enumerate(top_3.iterrows(), 1):
        print(f"   {idx}. {book['Title']} by {book['Author']} ({book['Rating']}★)")
    
    print(f"\n📊 Genre Breakdown:")
    for genre, count in df['Genre'].value_counts().items():
        avg_rating = df[df['Genre'] == genre]['Rating'].mean()
        print(f"   {genre}: {count} books (avg rating: {avg_rating:.2f}★)")
    
    print("\n" + "="*50)

generate_book_report(books_df)
```
